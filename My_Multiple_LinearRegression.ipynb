```python
import pandas as pd
import numpy as np
```

```python
class MyMultipleRegression():
  def __init__(self):
    self.coef_=None
    self.intercept_=None
  
  def fit(self,xTrain,yTrain):
    xTrain=np.insert(xTrain,0,1,axis=1)
    beta_values=np.linalg.inv(np.dot(xTrain.T,xTrain)).dot(xTrain.T).dot(yTrain)
    self.intercept_=beta_values[0]
    self.coef_=beta_values[1:]

  def predict(self,xTest):
    prediction=np.dot(xTest,self.coef_)+self.intercept_
    return np.array(prediction)
```

```python
from sklearn.datasets import load_diabetes
df=load_diabetes()
x=pd.DataFrame(df.data,columns=df.feature_names)
y=pd.DataFrame(df.target,columns=['target'])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

xTrain,xTest,yTrain,yTest=train_test_split(x,y,test_size=0.2,random_state=42)
lr=LinearRegression()
```

```python
lr.fit(xTrain,yTrain)
y_pred=lr.predict(xTest)
print('r2: ',r2_score(yTest,y_pred))
print(lr.coef_)
print(lr.intercept_)
```

```python
mmlr = MyMultipleRegression()
mmlr.fit(xTrain,yTrain)
y2_pred=mmlr.predict(xTest)
print("r2: ",r2_score(yTest,y2_pred))
```

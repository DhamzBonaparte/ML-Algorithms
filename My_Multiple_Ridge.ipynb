```python
from sklearn.linear_model import LinearRegression
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_diabetes
```

```python
x=load_diabetes().data
y=load_diabetes().target
```

```python
xTrain,xTest,yTrain,yTest = train_test_split(x,y,test_size=0.2,random_state=4)
```

### Using Ridge to know the expected coefficients and intercepts

```python
from sklearn.linear_model import Ridge
r=Ridge(alpha=0.1,solver='cholesky')
r.fit(xTrain,yTrain)
y_pred=r.predict(xTest)
print(r.coef_,r.intercept_)
```

```python
from sklearn.metrics import r2_score
r2_score(yTest,y_pred)
```

```python
class MyMultipleRidge():
  def __init__(self,alpha):
    self.alpha=alpha
    self.coef_=None
    self.intercept_=None
  
  def fit(self,xTrain,yTrain):
    xTrain=np.insert(xTrain,0,1,axis=1)
    identity = np.identity(xTrain.shape[1])
    res = np.linalg.inv(np.dot(xTrain.T,xTrain)+self.alpha*identity).dot(xTrain.T).dot(yTrain)
    self.coef_=res[1:]
    self.intercept_=res[0]
    return (self.coef_,self.intercept_)

  def predict(self,xTest):
    return np.dot(xTest,self.coef_)+self.intercept_
```

```python
mmr=MyMultipleRidge(0.1)
mmr.fit(xTrain,yTrain)
```

### Since the result of our MyMultipleRidge class is not as the ones of Ridge class this is because the "cholesky" solver keeps the first element of identity matrix as 0 because it is the column of our intercept and we dont want the slope values to affect the intercept value 

```python
r2_score(yTest,mmr.predict(xTest))
```

```python
class MyMultipleRidge():
  def __init__(self,alpha):
    self.alpha=alpha
    self.coef_=None
    self.intercept_=None
  
  def fit(self,xTrain,yTrain):
    xTrain=np.insert(xTrain,0,1,axis=1)
    identity = np.identity(xTrain.shape[1])
    identity[0][0]=0
    res = np.linalg.inv(np.dot(xTrain.T,xTrain)+self.alpha*identity).dot(xTrain.T).dot(yTrain)
    self.coef_=res[1:]
    self.intercept_=res[0]
    return (self.coef_,self.intercept_)

  def predict(self,xTest):
      return np.dot(xTest,self.coef_)+self.intercept_
```

```python
mmr=MyMultipleRidge(0.1)
mmr.fit(xTrain,yTrain)
```

### Now we have the expected answer after applying identity[0][0]=0

```python
r2_score(yTest,mmr.predict(xTest))
```

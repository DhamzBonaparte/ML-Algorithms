```python
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import load_diabetes
val=load_diabetes()
df=pd.DataFrame(val.data,columns=[val.feature_names])
```

```python
df['target']=val.target
x=df.drop(columns='target')
y=df['target']
```

```python
from sklearn.model_selection import train_test_split
xTrain,xTest,yTrain,yTest=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(xTrain,yTrain)
y_pred=lr.predict(xTest)
```

```python
from sklearn.metrics import r2_score
print("r2: ",r2_score(yTest,y_pred))
```

```python
print(lr.coef_,lr.intercept_)
```

# Always remember to flatten your ytrain data since dot product is happeneing, the shape of matrix can affect the result


```python
class MyMultipleGDRegressor():
  def __init__(self,epoch,learning_rate):
    self.epoch = epoch
    self.lr=learning_rate
    self.intercept_=None
    self.coef_=None

  def fit(self,xTrain,yTrain):
    self.intercept_=0
    self.coef_= np.ones(xTrain.shape[1])
    X = xTrain.values
    y = yTrain.values.reshape(-1)

    for i in range(self.epoch):
      yHat=np.dot(X,self.coef_)+self.intercept_
      slope_b0=-2*np.mean(y-yHat)
      self.intercept_=self.intercept_-(self.lr*slope_b0)
      
      error=y-yHat
      der_coef=-2 * np.dot((error),X)/X.shape[0]
      self.coef_=self.coef_-(self.lr*der_coef)

  def predict(self,xTest):
    return np.dot(xTest,self.coef_)+self.intercept_
```

```python
mgr=MyMultipleGDRegressor(400,0.8)
mgr.fit(xTrain,yTrain)
```

```python
y_pred_2=mgr.predict(xTest)
```

```python
from sklearn.metrics import r2_score
print("r2: ",r2_score(yTest,y_pred_2))
```

As we can see here that it is taking 400 epochs to get a desirable result which is not ideal for large datasets. To overcome this we use Stochastic Gradient Descent

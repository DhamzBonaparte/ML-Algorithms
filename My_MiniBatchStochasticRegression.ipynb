```python
import numpy as np
import pandas as pd
import random
```

```python
from sklearn.datasets import load_diabetes
val=load_diabetes()
df=pd.DataFrame(val.data,columns=[val.feature_names])
```

```python
df['target']=val.target
x=df.drop(columns='target')
y=df['target']
```

```python
from sklearn.model_selection import train_test_split
xTrain,xTest,yTrain,yTest=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
class My_Mini_Batch_Stochastic_Gradient_Descent():
    def __init__(self,epoch,learning_rate,batch_size):
      self.intercept_=None
      self.coef_=None
      self.epoch=epoch
      self.lr=learning_rate
      self.batch=batch_size
    
    def fit(self,xTrain,yTrain):
      self.intercept_=0
      self.coef_=np.ones(xTrain.shape[1])
      x=xTrain.values
      y=yTrain.values.reshape(-1)
     
      for i in range(self.epoch):
        for bat in range(int(xTrain.shape[0]/ self.batch)):
          
          idx = random.sample(range(xTrain.shape[0]),self.batch)
          yHat=np.dot(x[idx],self.coef_)+self.intercept_

          coef_der=-2*np.dot((y[idx]-yHat),x[idx])
          self.coef_=self.coef_-(self.lr*coef_der)

          slope_b0=-2*np.mean(y[idx]-yHat)
          self.intercept_=self.intercept_-(self.lr*slope_b0)
      
      return (self.intercept_,self.coef_)

    def predict(self,xTest):
      return np.dot(xTest,self.coef_)+self.intercept_
      
```

```python
mm=My_Mini_Batch_Stochastic_Gradient_Descent(100,0.01,int(xTrain.shape[0]/5))
mm.fit(xTrain,yTrain)
```

```python
y_pred=mm.predict(xTest)
```

```python
from sklearn.metrics import r2_score
print("r2: ",r2_score(yTest,y_pred))
```

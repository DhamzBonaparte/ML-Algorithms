```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,Y=make_regression(n_samples=100,n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)
```

```python
from sklearn.model_selection import train_test_split
xTrain,xTest,yTrain,yTest=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(xTrain,yTrain)
y_pred=lr.predict(xTest)
```

```python
from sklearn.metrics import r2_score
print("R2:",r2_score(yTest,y_pred))
```

# **For finding b-intercept**

```python
class GDRegressor():
  def __init__(self,learning_rate,epochs):
    self.b=0
    self.m=28.164
    self.lr=learning_rate
    self.epochs=epochs
  
  def fit(self,xTrain,yTrain):
    for i in range(self.epochs):
      loss_func=-2*np.sum(yTrain-self.m*xTrain.ravel()-self.b)
      self.b=self.b-(self.lr*loss_func)
    print(self.b)
```

```python
gdr=GDRegressor(0.1,10)
gdr.fit(xTrain,yTrain)
```

There is a very high diffrence between expected answer and what we actually got this is because our learnng rate is way too high:

```python
gdr=GDRegressor(0.01,10)
gdr.fit(xTrain,yTrain)
```

Now as we can see it is closer to the expexted b intercept (-2.71) but still not equal now we increase the epochs:

```python
gdr=GDRegressor(0.01,100)
gdr.fit(xTrain,yTrain)
```

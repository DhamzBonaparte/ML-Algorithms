```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
np.random.seed(42)
X = np.random.rand(100, 1) - 0.5
y = 3*X[:, 0]**2 + 0.05 * np.random.randn(100)
df = pd.DataFrame(X, columns=['X'])
df['Y'] = y
```

```python
df['pred1']=df['Y'].mean()
df['err']=df['Y']-df['pred1']
df
```

```python
from sklearn.tree import DecisionTreeRegressor
dtc=DecisionTreeRegressor(max_leaf_nodes=8)
dtc.fit(df.iloc[:,[0]],df.iloc[:,3])
```

```python
df['pred2']=df['pred1']+dtc.predict(df.iloc[:,[0]])
```

```python
df['err2']=df['Y']-df['pred2']
df
```

```python
dtc2=DecisionTreeRegressor(max_leaf_nodes=8)
dtc2.fit(df.iloc[:,[0]],df.iloc[:,-1])
```

```python
df['pred3']=df['pred1']+df['pred2']+dtc2.predict(df.iloc[:,[0]])
```

```python
import numpy as np
class MyLR():
  def __init__(self):
    self.m=None
    self.c=None

  def fit(self,xTrain,yTrain):
    num=0
    den=0
    for x in range(xTrain.shape[0]):
      num = num + ((xTrain.iloc[x]-xTrain.mean()))*((yTrain.iloc[x]-yTrain.mean()))
      den = den + (xTrain.iloc[x]-xTrain.mean())**2

    self.m = num/den
    self.c = (yTrain.mean())-(self.m*xTrain.mean())

    print(self.m,self.c)

  def predict(self,xTest):
    ans_arr=[]
    for x in range(xTest.shape[0]):
      result=(self.m*xTest.iloc[x])+(self.c)
      ans_arr.append(result)

    return np.array(ans_arr)
```

```python
lr=MyLR()
```

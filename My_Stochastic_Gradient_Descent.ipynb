```python
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import load_diabetes
val=load_diabetes()
df=pd.DataFrame(val.data,columns=[val.feature_names])
```

```python
df['target']=val.target
x=df.drop(columns='target')
y=df['target']
```

```python
from sklearn.model_selection import train_test_split
xTrain,xTest,yTrain,yTest=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(xTrain,yTrain)
y_pred=lr.predict(xTest)
```

```python
from sklearn.metrics import r2_score
print("r2: ",r2_score(yTest,y_pred))
```

```python
print(lr.coef_,lr.intercept_)
```

```python
class MyStochasticGradientDescent():
  def __init__(self,epoch,learning_rate):
    self.epoch=epoch
    self.lr=learning_rate
    self.coef_=None
    self.intercept_=None

  def fit(self,xTrain,yTrain):
    self.intercept_= 0
    self.coef_=np.ones(xTrain.shape[1])
    x=xTrain.values
    y=yTrain.values

    for i in range(self.epoch):
      for j in range(xTrain.shape[0]):

        idx = np.random.randint(0,xTrain.shape[0])

        yHat=self.intercept_+np.dot(self.coef_,x[idx])
        slope_b0=-2*(y[idx]-yHat)
        self.intercept_=self.intercept_-(self.lr*slope_b0)

        coef_slope=-2*(y[idx]-yHat)*x[idx]
        self.coef_=self.coef_-(self.lr*coef_slope)
      
    return (self.coef_,self.intercept_)

  def predict(self,xTest):
    return (np.dot(xTest,self.coef_)+self.intercept_)
  
```

```python
msgd=MyStochasticGradientDescent(20,0.1)
msgd.fit(xTrain,yTrain)
```

```python
y_pred=msgd.predict(xTest)
```

```python
print("r2_score: ", r2_score(yTest,y_pred))
```

```python
from sklearn.linear_model import SGDRegressor
sgd=SGDRegressor(max_iter=27,learning_rate='constant',eta0=0.1)
sgd.fit(xTrain,yTrain)
y_pred=sgd.predict(xTest)
r2_score(yTest,y_pred)
```

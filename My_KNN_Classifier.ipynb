```python
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
X = data.data
y = data.target
```

```python
from sklearn.model_selection import train_test_split
xTrain, xTest, yTrain, yTest = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from collections import Counter
kn = KNeighborsClassifier(n_neighbors=5)
kn.fit(xTrain, yTrain)
y_pred= kn.predict(xTest)
accuracy_score(yTest, y_pred)
```

```python
class MyKNN:
    def __init__(self,n_neighbors=5):
        self.n_neighbors = n_neighbors
    
    def fit(self,xTrain,yTrain):
        self.xTrain = xTrain
        self.yTrain = yTrain    
    
    def predict(self,xTest):
        yPred=[]
        for i in xTest:
            arr=[]
            for j in self.xTrain:
                arr.append(self.calcDistance(i,j))
            top=sorted(list(enumerate(arr)),key=lambda x:x[1])[0:self.n_neighbors]
            val=self.counted(top)
            yPred.append(val)
        return yPred
    
    def calcDistance(self, x1, x2):
        return np.linalg.norm(x1-x2)
    
    def counted(self,top):
        val=[]
        for i in top:
            val.append(self.yTrain[i[0]])
        val=Counter(val)
        return val.most_common()[0][0]
```

```python
knn=MyKNN(n_neighbors=5)
knn.fit(xTrain,yTrain)
knn.predict(xTest)
```

```python
accuracy_score(yTest, knn.predict(xTest))
```
